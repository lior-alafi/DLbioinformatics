```python
# import plotly.express as px
import matplotlib.pyplot as plt
import numpy as np

```

```python
training_seq_lengths = []
training_RBPs2_lengths= []
n_rna_samples = 0 
n_RBPs2_samples = 0
```

```python
with open('data/training_seqs.txt', 'r') as file:
    for line in file.readlines():
        line = line.strip()
        if len(line) > 0:
            n_rna_samples += 1
        training_seq_lengths.append(len(line))

with open('data/training_RBPs2.txt', 'r') as file:
    for line in file.readlines():
        line = line.strip()
        if len(line) > 0:
            n_RBPs2_samples += 1
        training_RBPs2_lengths.append(len(line))
```

```python
# px.histogram(training_seq_lengths, title='Distribution of RNA Sequence Lengths in Training Set')
```

```python
plt.hist(training_seq_lengths, edgecolor='black')  # can also use 'counts' and 'bin_edges' directly if needed
plt.title('Distribution of RNA Sequence Lengths in Training Set')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
#px.histogram(training_RBPs2_lengths, title='Distribution of RBPs2 Sequence Lengths in Training Set')
```

```python
plt.hist(training_RBPs2_lengths, edgecolor='black')  # can also use 'counts' and 'bin_edges' directly if needed
plt.title('Distribution of RBPs2 Sequence Lengths in Training Set')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
print(f'max RNA length: {max(training_seq_lengths)} rna seqs: {n_rna_samples}\nmax RBPs2 length: {max(training_RBPs2_lengths)} rbp2 seqs: {n_RBPs2_samples}\n')
```

```python
count = 0
train_res_lengths = []
y_per_rbp2 = []

with open('data/training_data2.txt','r') as file:
    for line in file.readlines():
        line = line.strip()
        if len(line) > 0:
            count += 1
        arr = line.split('\t')
        if count == 0:
            print(arr)
        train_res_lengths.append(len(arr))
        y_per_rbp2.append(np.array(arr).astype(np.float32))
```

```python
# px.histogram(train_res_lengths, title='Distribution of training result Sequence Lengths in Training Set')
```

```python
plt.hist(train_res_lengths, edgecolor='black')  # can also use 'counts' and 'bin_edges' directly if needed
plt.title('Distribution of training result Sequence Lengths in Training Set')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python


print(f'max training result length: {max(train_res_lengths)} avg {np.mean(train_res_lengths)}')
print(f'max values {np.max(y_per_rbp2)}')
print(f'min values {np.min(y_per_rbp2)}')
print(f'avg values {np.mean(y_per_rbp2)}')
```
